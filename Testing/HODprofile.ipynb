```python
def F_fun(x):
    return (1 + x*x)**1.5
def G_fun(y):
    return np.exp(-y)

x = np.logspace(-3, 3, num=60, endpoint=False)
F = F_fun(x)
H = mcfit.GaussVar(x, lowring=True)
y, G = H(F, extrap=True)
np.allclose(G, G_fun(y), rtol=1e-8, atol=1e-8)

H2 = mcfit.GaussVar(y, lowring=True)
x2, F2 = H2(G, extrap=True)


fig, axs = plt.subplots(1, 2, figsize=(15, 5), layout='constrained', sharex=True)
axs = axs.flatten()

axs[0].plot(x, F, ls='--')
axs[1].plot(y, G)
axs[0].plot(x2, F2, ls='--')


axs[0].set(ylabel=r'$\hat{n}_s (r, z, M_h) \ [\text{Mpc}^{-3}]$', xlabel=r'$r \ [\text{Mpc}]$', xscale='log', yscale='log'); axs[0].legend()
axs[1].set(ylabel=r'$\hat{n}_s (k, z, M_h) \ [\text{Mpc}^{-3}]$', xlabel=r'$k \ [\text{Mpc}^{-1}]$', yscale='log'); axs[1].legend()

plt.show()
```
