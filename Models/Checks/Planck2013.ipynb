```python
importlib.reload(Planck2013)
fig, axs = Planck2013.Fig4().full()

from Models import HaloModels
from Models.Papers import Arnaud2010
importlib.reload(HaloModels); importlib.reload(Arnaud2010)

r = np.geomspace(1e-3, 1e1, 50)
z, logM500c = 0.2, np.log10(2e13)
hmod = HaloModels.pyccl_model(Planck2013.Planck2013().info)
x = r*u.Mpc/hmod.rdel('500c')(z, logM500c)

P_del = lambda P13: (P13.Pressure(r*0.7**2, z, logM500c)()/P13.P500(z, logM500c))[:, 0, 0]

P13 = lambda **spefs: Planck2013.Planck2013(**spefs, H=hmod.H, rhoc=hmod.rhoc, r500c=hmod.rdel('500c'))
axs[0].scatter(x, P_del(P13(cluster='All',fixedp='1')), marker='o', alpha=0.5, c='w', label='P13 all')
axs[1].scatter(x, P_del(P13(cluster='cool')), marker='o', alpha=0.5, c='w', label='P13 cool')   
axs[1].scatter(x, P_del(P13(cluster='noncool')), marker='x', alpha=0.5, c='w', label='P13 noncool')                                                                            

A10 = lambda **spefs: Arnaud2010.HaloProfiles(**spefs, H=hmod.H, rhoc=hmod.rhoc, r500c=hmod.rdel('500c'))
axs[0].scatter(x, P_del(A10(model='norm')), marker='o', alpha=0.5, c='g', label='A10')
axs[1].scatter(x, P_del(A10(model='norm')), marker='o', alpha=0.5, c='g', label='A10')                                                     

axs[0].legend(); axs[1].legend()
plt.show()
```
